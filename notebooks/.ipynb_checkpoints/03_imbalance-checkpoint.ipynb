# 03. Handling Class Imbalance

**Question:** does changing how we train (class weights, SMOTE, undersampling) or where we cut (threshold tuning) improve fraud detection, and by how much?

**Fair comparison:** every strategy uses the **same logistic regression**, the **same features**, the **same time-based split**, and the **same evaluation protocol** as the baseline. Only the imbalance strategy changes.

| Strategy | What it does | Risk |
|---|---|---|
| **Unweighted** (baseline) | Nothing special | Model rarely scores fraud above 0.5, so use a tuned threshold |
| **Class weights** | Each fraud counts about as much as all legit rows combined, so errors on fraud cost more in training | Probabilities become much too high |
| **SMOTE (10%)** | Creates synthetic fraud rows by interpolating between real fraud neighbors, until fraud = 10% of legit | Synthetic points can be unrealistic; probabilities shift |
| **Undersampling (10%)** | Throws away legit rows until fraud = 10% of legit | Discards information |
| **Threshold tuning** | Not a training change. Move the cut-off on validation | Needs a validation set with enough frauds |

**Leakage rules enforced in the code:** resampling is inside the training pipeline and runs only during `fit` on train. Validation and test are never resampled, so their precision and recall stay honest.

## 2. Ranking quality (threshold-free)
PR-AUC and recall at 90% precision judge the **ranking** of transactions, independent of any threshold.

```python
rank
```

**How to read it**
- **95% CI** is the bootstrap range for PR-AUC. With few frauds in test the ranges are wide.
- **diff vs unweighted** is a *paired* bootstrap difference. If the range includes 0, the strategy is **not distinguishable** from the baseline on this test set. Only trust a difference whose range excludes 0.
- **Expectation:** class weights, SMOTE and undersampling mostly change *probabilities*, not the *ranking*, so PR-AUC often barely moves (or drops slightly). Check whether yours does.
- **mean score / Brier:** the true fraud rate is the benchmark for the mean score. A much higher mean score means the probabilities are inflated (see section 5).

**How to read it**

- Counts matter: a difference of a couple of frauds is noise with this few positives.

## 5. An illustrative cost comparison
A simple preview of step 3. **These costs are assumptions for illustration, not facts:** edit them to match your own. Total cost = FN × cost of a miss + FP × cost of a false alarm.

```python
COST_FN = 100   # assumed cost of a missed fraud (flat; step 3 will use the transaction amount)
COST_FP = 5     # assumed cost of reviewing a false alarm

rows = []
for name, r in res.items():
    for label, key in [("0.5", "at_0.5"), ("val-tuned", "at_val_tuned")]:
        k = r[key]
        rows.append({"strategy": name, "threshold": label, "FN": k["fn"], "FP": k["fp"],
                     "total cost": k["fn"] * COST_FN + k["fp"] * COST_FP})
costs = pd.DataFrame(rows)
display(costs.pivot(index="strategy", columns="threshold", values="total cost").loc[list(res)])
```

**How to read it:** compare costs *within the same threshold column* and *between columns*. If tuning the threshold cuts cost much more than switching strategy, threshold tuning is the cheaper, simpler lever. Change `COST_FN` and `COST_FP` and rerun to see how the best choice depends on your assumptions.

## 6. Probability distortion (calibration check)
Weighting and resampling teach the model that fraud is more common than it is, so scores no longer mean "probability of fraud". The test below compares the average predicted score to the true fraud rate.

```python
calib = pd.DataFrame({
    "mean score": {n: r["mean_score_test"] for n, r in res.items()},
    "true fraud rate": yte.mean(),
    "ratio": {n: r["mean_score_test"] / yte.mean() for n, r in res.items()},
}).round(4)
calib
```

**How to read it:** a ratio near 1 means well-calibrated scores. A ratio of 5 or 10 means scores are inflated by that factor. This matters because the cost-based threshold rule in step 3 (`flag if p ≥ C_FP / (C_FP + C_FN)`) only works with calibrated probabilities. If you use class weights or SMOTE, either calibrate the scores (isotonic or Platt, fit on validation) or choose thresholds by sweeping actual costs.
