# 04. Gradient-Boosted Trees and a Cost-Based Threshold

**Goals**
1. Train LightGBM and check whether it beats the logistic regression baseline (with uncertainty).
2. Report PR-AUC, recall at 90% precision, and recall at a fixed **alert budget**.
3. Choose the decision threshold using **money**, not F1.

**Protocol (unchanged):** time-based 70/15/15 split. Everything we *choose* (hyperparameters, early stopping, calibration, thresholds) uses **validation only**. The test set is used once, at the end.

### Cost assumptions (stated on purpose, edit them below)
| Event | Cost | Reasoning |
|---|---|---|
| **Missed fraud (FN)** | the transaction **Amount** | The money is lost. Ignores chargeback recovery |
| **False alarm (FP)** | a flat **$C_FP** (default $5) | Analyst review time plus customer friction |
| Caught fraud / correct pass | $0 | The loss is prevented / nothing happened |

A missed $500 fraud therefore costs 100 times more than a missed $5 fraud, which a plain F1 threshold ignores.

```python
import sys
sys.path.insert(0, "../src")
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import Image, display

from fraud.data import load_raw, make_synthetic
from fraud.gbt import run_gbt_experiment
from fraud.reporting import gbt_figures, gbt_tables

# Set True ONLY to check that the notebook runs without the real file. Never quote synthetic results.
USE_SYNTHETIC = False
DATA_PATH = Path("../data/creditcard.csv")
FIG = Path("../docs/figures"); FIG.mkdir(parents=True, exist_ok=True)
plt.rcParams.update({"figure.dpi": 100, "axes.grid": True, "grid.alpha": 0.3})

df = make_synthetic() if USE_SYNTHETIC else load_raw(DATA_PATH)
if USE_SYNTHETIC:
    print("WARNING: SYNTHETIC DATA. Code-path check only.")
df.shape
```

## 1. Settings

```python
C_FP = 5.0      # assumed cost of a false alarm, in dollars. Try 1, 5, 20 (section 8 shows the sensitivity)
N_ITER = 12     # random-search trials for LightGBM (more = better search, slower)
```

## 2. Train: random search with early stopping
Each trial trains up to 1000 trees and stops when validation PR-AUC hasn't improved for 50 rounds. The best trial (by **validation** PR-AUC) wins. Trials rotate through three `scale_pos_weight` settings so we can see whether imbalance weighting helps trees (the same question as step 2).

```python
out = run_gbt_experiment(df, c_fp=C_FP, n_iter=N_ITER)
yte = out["y_test"]
print(f"train: {out['train_rows']:,} rows ({out['train_fraud']} fraud) | test: {len(yte):,} rows ({int(yte.sum())} fraud, {yte.mean():.3%})")
print("best params:", out["best_params"], "| best iteration:", out["best_iteration"])
rank, costs, sens, cal = gbt_tables(out)
out["trials"].head(10)
```

```python
# Does imbalance weighting help trees? Best validation PR-AUC per scale_pos_weight setting
out["trials"].groupby("spw_name")["val_PR_AUC"].agg(["max", "mean", "count"]).round(4).sort_values("max", ascending=False)
```

**How to read it:** the search is small and the validation set has few frauds, so differences of a point or two between trials are noise. If `none (1)` is close to the weighted settings, prefer it (simpler, and better-calibrated probabilities).

## 4. Calibration
Class weighting (and boosting in general) can distort probabilities. The cost rule in section 6 needs scores that mean "probability of fraud", so we calibrate with **Platt scaling** fitted on validation (a small logistic regression on the model's score). It does not change the *ranking*, only the scale.

```python
cal
```

**How to read it:** compare *mean score* with the *true fraud rate*. After calibration they should be close, and the Brier score (lower is better) should improve or stay similar. Calibration is fit on a validation set with few frauds, so treat it as approximate.

## 5. Decisions and their cost (test)
Each row is a different decision rule, evaluated once on test with the cost assumptions above. **Savings** compares each rule with flagging nothing.

| Rule | How the threshold is chosen |
|---|---|
| Flag nothing / everything | Reference points |
| LR + cost threshold | The baseline model with the same cost-based threshold method |
| LightGBM, 0.5 | The naive default |
| LightGBM, F1-tuned | Maximizes F1 on validation (ignores amounts) |
| LightGBM, **cost-tuned** | Minimizes total cost on validation using real amounts |
| LightGBM, **amount-aware** | Flag if `p >= C_FP / (C_FP + Amount)`, using calibrated probabilities: big transactions get a lower bar |

```python
costs
```

**How to read it**
- **Total cost = missed $ + review $.** The best rule balances the two.
- The amount-aware rule comes from expected loss: flag when `p × Amount ≥ (1 − p) × C_FP`. It uses a *different threshold per transaction*, so it needs good calibration. If it is worse than the single cost-tuned threshold, calibration is probably the weak link.
- Few test frauds means cost differences between rules can come from one or two big transactions. Check the TP/FP/FN columns before concluding.

Validation (solid) and test (dashed) should have similar shapes. If the validation minimum is far from the test minimum, the chosen threshold is not very stable.

## 7. Save the final model (real data only)
Saves the preprocessing + LightGBM **pipeline** and the calibrator, so the API in step 5 uses exactly the code that was evaluated.

```python
import json, joblib
if not USE_SYNTHETIC:
    models = Path("../models"); models.mkdir(exist_ok=True)
    joblib.dump(out["pipeline"], models / "lgbm_pipeline.joblib")
    joblib.dump(out["calibrator"], models / "platt_calibrator.joblib")
    (models / "config.json").write_text(json.dumps(
        {"c_fp": C_FP, "threshold_cost_tuned": out["thresholds"]["cost"],
         "threshold_f1_tuned": out["thresholds"]["f1"], "best_params": out["best_params"]}, indent=2))
    print("saved to", models.resolve())
else:
    print("synthetic data: nothing saved")
```
