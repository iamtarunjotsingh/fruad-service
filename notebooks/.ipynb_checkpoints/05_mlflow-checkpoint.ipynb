# 05. Experiment Tracking with MLflow

**Problem:** after steps 1 to 3 you have many results scattered across notebooks. Which model was best? What exact settings produced it? Can you reproduce it next month? MLflow answers this by recording every run.

**Vocabulary**
| Term | Meaning | In this project |
|---|---|---|
| **Experiment** | A named group of runs | `fraud-detection` |
| **Run** | One execution that is recorded | One model fit: majority class, each imbalance strategy, each LightGBM trial, the final model |
| **Parameters** | Settings you chose | model, strategy, seed, `num_leaves`, false-alarm cost |
| **Metrics** | Numbers you measured | `test_pr_auc`, `test_recall_at_90_precision`, cost savings |
| **Tags** | Labels for context | git commit, data fingerprint, library versions |
| **Artifacts** | Files attached to a run | figures, result tables, the saved model, the calibrator |
| **Nested runs** | Child runs inside a parent | the 12 search trials under "lightgbm search" |

**Where the data lives:** a local SQLite file `mlflow.db` plus an `mlartifacts/` folder in the project. No server is needed. Naming convention: metrics that come from the **test** set start with `test_`, validation ones with `val_`, so nobody confuses them.

````python
import sys
sys.path.insert(0, "../src")
from pathlib import Path

import matplotlib.pyplot as plt
import mlflow
import numpy as np
import pandas as pd
from mlflow.tracking import MlflowClient

from fraud.data import DEFAULT_PATH, load_raw, make_synthetic
from fraud.tracking import leaderboard, load_logged_model, track_experiments

# Set True ONLY to check that the notebook runs without the real file. Synthetic runs go to a SEPARATE database.
USE_SYNTHETIC = False
ROOT = Path("..").resolve()
DB = "mlflow_synthetic.db" if USE_SYNTHETIC else "mlflow.db"
TRACKING_URI = f"sqlite:///{(ROOT / DB).as_posix()}"
ARTIFACT_ROOT = ROOT / ("mlartifacts_synthetic" if USE_SYNTHETIC else "mlartifacts")
plt.rcParams.update({"figure.dpi": 100, "axes.grid": True, "grid.alpha": 0.3})

df = make_synthetic() if USE_SYNTHETIC else load_raw(ROOT / "data" / "creditcard.csv")
if USE_SYNTHETIC:
    print("WARNING: SYNTHETIC DATA. Code-path check only.")
df.shape
````

## 1. Run everything with tracking
This re-runs the baselines, the imbalance comparison and the LightGBM search, and records each as MLflow runs. It takes a few minutes on the real data. Set `RUN = False` later to just browse existing results.

````python
RUN = True
C_FP = 5.0      # assumed cost of a false alarm (dollars)
N_ITER = 12     # LightGBM search trials

if RUN:
    result = track_experiments(df, data_path=None if USE_SYNTHETIC else ROOT / "data" / "creditcard.csv",
                               synthetic=USE_SYNTHETIC, c_fp=C_FP, n_iter=N_ITER,
                               tracking_uri=TRACKING_URI, artifact_root=ARTIFACT_ROOT)
    FINAL_RUN_ID = result["run_ids"]["lightgbm_final"]
    print("final LightGBM run:", FINAL_RUN_ID)
````

## 2. Leaderboard
Every model with a test metric, best first (latest run per name).

````python
lb = leaderboard(TRACKING_URI)
lb.drop(columns="run_id").round(4)
````

**How to read it:** this one table replaces hunting through notebooks. Remember the lessons from earlier steps: with about 50 test frauds, small PR-AUC gaps are within noise (see the bootstrap intervals logged as `test_pr_auc_ci_low/high`). `cost savings %` exists only for the final LightGBM run because that is where the cost analysis lives.

**Your analysis:** _write here_

````python
fig, ax = plt.subplots(figsize=(8, 3.8))
top = lb.sort_values("PR-AUC")
ax.barh(top["run"], top["PR-AUC"], color="#4c78a8")
ax.axvline(float(df["Class"].mean()), ls="--", c="grey", label="random (fraud rate)")
ax.set_xlabel("test PR-AUC"); ax.set_title("All tracked models"); ax.legend()
fig.tight_layout(); plt.show()
````

## 3. Look inside one run
Everything needed to reproduce the final LightGBM model is attached to its run.

````python
client = MlflowClient(TRACKING_URI)
run = client.get_run(FINAL_RUN_ID)
print("PARAMS"); display(pd.Series(run.data.params, name="value").to_frame())
print("METRICS"); display(pd.Series(run.data.metrics, name="value").round(4).to_frame())
print("TAGS (reproducibility)")
display(pd.Series({k: v for k, v in run.data.tags.items() if not k.startswith("mlflow.")}, name="value").to_frame())
````

````python
for sub in ["", "report", "calibrator", "model"]:
    print(f"[{sub or 'root'}]", [f.path for f in client.list_artifacts(FINAL_RUN_ID, sub or None)][:12])
````

**How to read it:**
- **Params** are inputs: if you re-run with the same params, seed and data, you should get the same metrics.
- **Tags** pin down *which code* (`git_commit`, `git_dirty`) and *which data* (an md5 fingerprint of `creditcard.csv`) produced the run. `git_dirty = True` means there were uncommitted changes, so the commit alone does not capture the code.
- **Artifacts** hold the figures, result tables, the model and the calibrator.

## 4. Which hyperparameters mattered? (the search trials)
The 12 trials are child runs. Querying them shows how each setting related to validation PR-AUC.

````python
trials = mlflow.search_runs(experiment_names=["fraud-detection"], filter_string="tags.stage = 'lightgbm-trial'")
cols = [c for c in trials.columns if c.startswith("params.") and c not in ("params.model",)]
t = trials[["tags.mlflow.runName", "metrics.val_pr_auc", "metrics.best_iteration"] + cols].rename(columns=lambda c: c.split(".")[-1])
for c in t.columns[1:]:      # convert numeric-looking columns; leave text columns (like spw_name) alone
    conv = pd.to_numeric(t[c], errors="coerce")
    if conv.notna().sum() == t[c].notna().sum():
        t[c] = conv
t.sort_values("val_pr_auc", ascending=False).head(10).round(4)
````

````python
num = [c for c in ["num_leaves", "min_child_samples", "subsample", "colsample_bytree", "reg_lambda", "scale_pos_weight"] if c in t.columns]
fig, axes = plt.subplots(1, len(num), figsize=(3 * len(num), 3), sharey=True)
for ax, c in zip(np.atleast_1d(axes), num):
    ax.scatter(pd.to_numeric(t[c], errors="coerce"), t["val_pr_auc"]); ax.set_xlabel(c)
np.atleast_1d(axes)[0].set_ylabel("validation PR-AUC")
fig.tight_layout(); plt.show()
````

**How to read it:** each dot is one trial. A clear trend (for example higher `scale_pos_weight` consistently worse) is a finding. A cloud with no trend means that setting barely matters, so keep the simpler value. With only 12 trials and few validation frauds, expect mostly clouds.

## 5. Reload the logged model
The model is saved as one pipeline (preprocessing + LightGBM), so anyone can load it by run id and get exactly what was evaluated. This is what the API will do in step 5.

````python
model = load_logged_model(FINAL_RUN_ID, TRACKING_URI)
sample = df.drop(columns="Class").tail(5)
pd.DataFrame({"fraud_score": model.predict_proba(sample)[:, 1].round(5), "actual": df["Class"].tail(5).to_numpy()})
````

**Note on loading models:** the model is stored with `cloudpickle`, which means loading it runs code from the file. Only load models you or your team created. The `fraud` package must be importable (run `pip install -e .` from the project root once).

## 6. Browse in the MLflow UI
In a terminal at the project root:

```
mlflow ui --backend-store-uri sqlite:///mlflow.db
```
then open http://127.0.0.1:5000. (If `mlflow` is not found, use `python -m mlflow ui ...`.)

Things to try:

2. Sort the table by `test_pr_auc` or `cost_savings_pct`.
3. Open the final run, then **Artifacts** to see the PR-curve figure, the cost curve and the CSV tables.
4. Expand the "lightgbm search" parent run to see its 12 child trials.
