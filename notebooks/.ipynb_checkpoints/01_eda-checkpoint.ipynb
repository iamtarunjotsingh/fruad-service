# 01. Exploratory Data Analysis: Card Fraud Detection

**Goal:** understand the data well enough to choose a split, metrics and features that won't mislead us.

**Dataset:** ULB credit card fraud. `Time` (seconds since first transaction), `Amount`, `Class` (1 = fraud),
and `V1..V28` (PCA outputs, so their original meaning is hidden).

**How to use this notebook:** run top to bottom. Each section has code, then a short *What this tells us* note.
Under each plot there is an **Your analysis** cell, where you write what you see in your own words.

```python
import sys
sys.path.insert(0, "../src")
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from fraud.data import TARGET, load_raw, make_synthetic, time_split, xy

# Set True ONLY to check that the notebook runs without the real file. Never quote synthetic results.
USE_SYNTHETIC = False
DATA_PATH = Path("../data/creditcard.csv")
FIG = Path("../docs/figures"); FIG.mkdir(parents=True, exist_ok=True)
plt.rcParams.update({"figure.dpi": 100, "axes.grid": True, "grid.alpha": 0.3})

df = make_synthetic() if USE_SYNTHETIC else load_raw(DATA_PATH)
if USE_SYNTHETIC:
    print("WARNING: SYNTHETIC DATA. Code-path check only.")
df.shape
```

## 1. Dataset overview and data quality

```python
overview = pd.Series({
    "rows": f"{len(df):,}",
    "columns": df.shape[1],
    "fraud rows": f"{int(df[TARGET].sum()):,}",
    "fraud rate": f"{df[TARGET].mean():.4%}",
    "legit : fraud": f"{(len(df) - df[TARGET].sum()) / df[TARGET].sum():,.0f} : 1",
    "missing values": int(df.isna().sum().sum()),
    "duplicate rows": f"{int(df.duplicated().sum()):,}",
    "time span (hours)": round((df["Time"].max() - df["Time"].min()) / 3600, 1),
}, name="value")
overview.to_frame()
```

**What this tells us**
- The publicly documented ULB data has about 285k rows and about 0.17% fraud. Check the table against that.
- A small number of exact duplicate rows is normal here. We keep them (they are genuine repeated transactions in a feed).
- No missing values means minimal cleaning. The work is in the imbalance, not in imputation.

## 2. Class balance
At this imbalance, accuracy is meaningless: always predicting "legit" would score about 99.8%.

```python
counts = df[TARGET].value_counts().sort_index()
fig, ax = plt.subplots(figsize=(5, 3.5))
ax.bar(["legit", "fraud"], counts.values, color=["#4c78a8", "#e45756"])
ax.set_yscale("log"); ax.set_ylabel("rows (log scale)"); ax.set_title("Class balance")
for i, v in enumerate(counts.values):
    ax.text(i, v, f"{v:,}", ha="center", va="bottom")
fig.tight_layout(); fig.savefig(FIG / "class_balance.png"); plt.show()
print(f"Accuracy of a model that never flags fraud: {1 - df[TARGET].mean():.4%}")
```

**What this tells us:** the "do-nothing" model looks excellent on accuracy and catches zero fraud.
**Decision:** evaluate with PR-AUC, recall at a fixed precision, and (later) a cost-based threshold.

## 3. Transaction amount

```python
print(df.groupby(TARGET)["Amount"].describe()[["count", "mean", "25%", "50%", "75%", "max"]]
      .rename(index={0: "legit", 1: "fraud"}).round(2))

fig, ax = plt.subplots(figsize=(7, 4))
bins = np.logspace(-2, np.log10(max(df["Amount"].max(), 10)), 50)
for c, label in [(0, "legit"), (1, "fraud")]:
    ax.hist(df.loc[df[TARGET] == c, "Amount"].clip(lower=0.01), bins=bins, alpha=0.6, density=True, label=label)
ax.set_xscale("log"); ax.set_xlabel("Amount (log scale)"); ax.set_ylabel("density")
ax.set_title("Amount distribution by class (normalized)"); ax.legend()
fig.tight_layout(); fig.savefig(FIG / "amount_by_class.png"); plt.show()
```

**How to read it:** amounts are heavily right-skewed. If the fraud curve sits left of the legit curve,
fraudsters favor small amounts (often card testing). If the curves overlap heavily, amount alone is a weak signal.
**Decision:** use `log1p(Amount)` as a feature. Keep raw `Amount` for the cost step, where a missed fraud costs more when the amount is larger.

## 4. Time of day

```python
hour = ((df["Time"] // 3600) % 24).astype(int)
by_hour = df.groupby(hour)[TARGET].agg(count="count", frauds="sum")
by_hour["fraud_rate_%"] = 100 * by_hour["frauds"] / by_hour["count"]

fig, ax = plt.subplots(figsize=(8, 3.8))
ax.bar(by_hour.index, by_hour["fraud_rate_%"])
ax.set_xlabel("hour of day (derived from Time)"); ax.set_ylabel("fraud rate (%)")
ax.set_title("Fraud rate by hour of day")
fig.tight_layout(); fig.savefig(FIG / "fraud_rate_by_hour.png"); plt.show()
by_hour.sort_values("fraud_rate_%", ascending=False).head(5).round(3)
```

**How to read it:** the data covers only two days, so each hour holds very few frauds and spikes may be chance. Treat this as a hint, not a finding.
**Decision:** encode hour as `sin/cos` so hour 23 sits next to hour 0. Do **not** feed raw `Time` to the model. It is an elapsed counter,
and a model that learns "fraud happens at second X" is useless on future data.

## 5. Which features separate fraud from legit?
The KS statistic compares the two distributions per feature (0 = identical, 1 = fully separated).

```python
from scipy.stats import ks_2samp

feats = [c for c in df.columns if c not in (TARGET, "Time")]
fraud, legit = df[df[TARGET] == 1], df[df[TARGET] == 0]
ks = pd.Series({f: ks_2samp(fraud[f], legit[f]).statistic for f in feats}).sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8, 3.8))
ks.head(15)[::-1].plot.barh(ax=ax, color="#4c78a8")
ax.set_xlabel("KS statistic"); ax.set_title("Top 15 features by class separation")
fig.tight_layout(); fig.savefig(FIG / "ks_top_features.png"); plt.show()

fig, axes = plt.subplots(1, 3, figsize=(12, 3.6))
for ax, f in zip(axes, ks.index[:3]):
    for c, label in [(0, "legit"), (1, "fraud")]:
        ax.hist(df.loc[df[TARGET] == c, f].clip(-15, 15), bins=60, alpha=0.6, density=True, label=label)
    ax.set_title(f"{f} (KS={ks[f]:.2f})"); ax.legend()
fig.tight_layout(); fig.savefig(FIG / "top_features.png"); plt.show()
```

**How to read it:** a high KS means one feature alone separates the classes well. A handful of V-features usually stand out on this dataset (commonly V14, V17, V12, V10), but confirm with your plot.
**Caveat for the demo:** V-features are anonymized, so SHAP later can only say "V14 pushed the score up". That is a limit of the dataset and worth stating in the README.

## 6. Time-based split
Sort by time. Train on the oldest 70%, validate on the next 15%, test on the newest 15%. A random split would leak same-hour fraud bursts across sets.

```python
tr, va, te = time_split(df)
split = pd.DataFrame({
    "rows": [len(d) for d in (tr, va, te)],
    "frauds": [int(d[TARGET].sum()) for d in (tr, va, te)],
    "fraud_rate_%": [round(100 * d[TARGET].mean(), 3) for d in (tr, va, te)],
    "start_h": [round(d["Time"].min() / 3600, 1) for d in (tr, va, te)],
    "end_h": [round(d["Time"].max() / 3600, 1) for d in (tr, va, te)],
}, index=["train", "validation", "test"])
display(split)

fig, ax = plt.subplots(figsize=(9, 3))
for d, name, col in [(tr, "train", "#4c78a8"), (va, "validation", "#f58518"), (te, "test", "#54a24b")]:
    f = d[d[TARGET] == 1]
    ax.scatter(f["Time"] / 3600, np.zeros(len(f)), s=14, color=col, label=f"{name} frauds")
    ax.axvspan(d["Time"].min() / 3600, d["Time"].max() / 3600, alpha=0.12, color=col)
ax.set_yticks([]); ax.set_xlabel("hours since start"); ax.set_title("Fraud events across the time split"); ax.legend(ncol=3)
fig.tight_layout(); fig.savefig(FIG / "time_split.png"); plt.show()
```

**What to check:** the number of frauds in validation and test. Only a few dozen frauds in test means each missed fraud moves recall by a couple of percentage points. Report TP/FP/FN counts next to rates, and avoid reading much into small differences between models.
A different fraud rate across splits is real drift, which is why time-based evaluation is the honest choice.

## Summary of decisions

| Finding | Decision |
|---|---|
| About 0.17% fraud | PR-AUC, recall at fixed precision, cost-based threshold; no accuracy |
| Amount is right-skewed | `log1p(Amount)` feature |
| `Time` is an elapsed counter | Replace with cyclic hour-of-day |
| Few frauds per split | Report counts; be careful about small metric differences |
| Clean data (no nulls) | Minimal cleaning; scale features, fit on train only |
